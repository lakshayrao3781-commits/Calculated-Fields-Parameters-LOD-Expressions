Question 1 :
What is a Calculated Field and why is it important in business dashboards?

Definition: A custom column created using mathematical formulas, logical statements, or Tableau functions on existing data fields.

Importance:

Enables dynamic KPI tracking (e.g., profit margins, delivery status) without altering the raw database.

Adds custom business logic and conditional formatting directly into dashboard visual elements.

Question 2 :
Why can’t parameters change visuals directly in Tableau?

Reason: Parameters are static input values (placeholders) that hold a single value at a time and do not possess inherent connections to data rows or views.

Workaround: They require a Calculated Field or Filter that references the parameter value to dynamically update the underlying data or alter visual outputs.

Question 3 :
When should EXCLUDE LOD be used?

Use Case: When you need to calculate an aggregate measure at a higher summary level while preserving lower-level granular dimensions in your visual display.

Example: Displaying a specific region’s overall average sales alongside individual product sales within that region without filtering out product details.

Question 4 :
Difference between FIXED and INCLUDE LOD expressions?

FIXED LOD:

Calculates aggregates using specified dimensions, ignoring the level of detail in the view.

Evaluated before standard dimension filters.

INCLUDE LOD:

Calculates aggregates using specified dimensions in addition to whatever dimensions are currently in the view.

Evaluated after standard dimension filters.

Task 5: Customer/Product Detail Impact (INCLUDE LOD)
Business Need
The sales team wants average sales considering product-level detail, even when product is not shown.
Task
Create a calculation that finds the average sales at the product level, but allows the result to be used at a
higher level (such as Category or Region). Use INCLUDE LOD.

AVG({ INCLUDE [Product_Name] : AVG([Sales]) })

Purpose: Forces Tableau to first aggregate average sales per product line before recalculating overall averages across higher levels (such as City or Category).
